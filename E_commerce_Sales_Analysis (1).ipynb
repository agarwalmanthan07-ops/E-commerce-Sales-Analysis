```python
# =======================================================
# E-commerce Sales Analysis (Day 8 Assignment)
# =======================================================

import io
import pandas as pd

# -------------------------------------------------------
# 1. Load Dataset
# -------------------------------------------------------
# Replace io.StringIO with actual CSV reading in Colab:
# df = pd.read_csv("ecommerce_sales.csv")

csv_data = """order_id,product,category,city,payment_method,sales,quantity
101,Laptop,Electronics,New York,Credit Card,1200.0,2
102,Chair,Furniture,Los Angeles,PayPal,150.0,1
103,Headphones,Electronics,Chicago,Credit Card,160.0,2
104,Desk,Furniture,New York,Debit Card,300.0,1
105,Phone,Electronics,Chicago,Credit Card,800.0,1
106,Monitor,Electronics,Los Angeles,PayPal,250.0,2
107,Sofa,Furniture,Chicago,Credit Card,500.0,1
108,Laptop,Electronics,New York,PayPal,1200.0,1
109,Headphones,Electronics,Los Angeles,Debit Card,80.0,1
110,Desk,Furniture,Chicago,Credit Card,300.0,3
"""

df = pd.read_csv(io.StringIO(csv_data))

print("=== Dataset Head ===")
print(df.head())

print("\n=== Dataset Info ===")
print(df.info())

# -------------------------------------------------------
# 2. Data Selection, Filtering, and Sorting
# -------------------------------------------------------
# Select key columns
sales_summary_cols = df[["order_id", "product", "sales", "quantity"]]
print("\n--- Selected Columns ---")
print(sales_summary_cols.head())

# Filter high-value sales (sales > 300)
high_value_orders = df[df["sales"] > 300]
print("\n--- High-Value Orders (> $300) ---")
print(high_value_orders)

# Sort by sales in descending order
sorted_by_sales = df.sort_values(by="sales", ascending=False)
print("\n--- Top Orders Sorted by Sales ---")
print(sorted_by_sales.head())

# -------------------------------------------------------
# 3. GroupBy & Aggregations
# -------------------------------------------------------
# Helper function for aggregated metrics
agg_metrics = {
    "sales": ["sum", "mean", "max", "min"],
    "quantity": "sum",
    "order_id": "count",
}

# Category Analysis
category_analysis = df.groupby("category").agg(agg_metrics)
category_analysis.columns = [
    "total_sales",
    "avg_sales",
    "highest_sale",
    "lowest_sale",
    "total_quantity",
    "order_count",
]

# City Analysis
city_analysis = df.groupby("city").agg(agg_metrics)
city_analysis.columns = [
    "total_sales",
    "avg_sales",
    "highest_sale",
    "lowest_sale",
    "total_quantity",
    "order_count",
]

# Product Analysis
product_analysis = df.groupby("product").agg(agg_metrics)
product_analysis.columns = [
    "total_sales",
    "avg_sales",
    "highest_sale",
    "lowest_sale",
    "total_quantity",
    "order_count",
]

# Payment Method Analysis
payment_analysis = df.groupby("payment_method").agg(agg_metrics)
payment_analysis.columns = [
    "total_sales",
    "avg_sales",
    "highest_sale",
    "lowest_sale",
    "total_quantity",
    "order_count",
]

print("\n=== Sales Analysis by Category ===")
print(category_analysis)

print("\n=== Sales Analysis by City ===")
print(city_analysis)

print("\n=== Sales Analysis by Product ===")
print(product_analysis)

print("\n=== Sales Analysis by Payment Method ===")
print(payment_analysis)

# -------------------------------------------------------
# 4. Top Performing Insights
# -------------------------------------------------------
top_product = product_analysis.sort_values(
    by="total_sales", ascending=False
).index[0]
top_category = category_analysis.sort_values(
    by="total_sales", ascending=False
).index[0]

print(f"\nTop Performing Category: {top_category}")
print(f"Top Performing Product : {top_product}")

# -------------------------------------------------------
# 5. Summary Observations & Findings
# -------------------------------------------------------
print("\n" + "=" * 50)
print("             KEY OBSERVATIONS & FINDINGS          ")
print("=" * 50)
print(
    """
1. Category Performance:
   Electronics is the highest revenue-generating category, driven largely by high-value sales of Laptops and Phones.

2. Product Performance:
   Laptops generate the highest total revenue among individual products, whereas smaller accessories like Headphones lead in total quantity sold.

3. Regional Breakdown:
   New York and Chicago demonstrate the highest transaction volumes and total revenue contributions across regions.

4. Payment Preferences:
   Credit Card transactions account for the largest share of overall volume and revenue compared to PayPal and Debit Cards.
"""
)
```
